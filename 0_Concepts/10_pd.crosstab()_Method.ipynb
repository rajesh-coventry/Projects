# **`pd.crosstab()`:**

`crosstab()` is a powerful function used to **compute a cross-tabulation of two or more factors** (usually categorical variables). It’s commonly used to **analyze the relationship between categorical features** in a dataset and is conceptually similar to a **pivot table** or **contingency table**.

```python
pandas.crosstab(index, columns, values=None, rownames=None, colnames=None, aggfunc=None, margins=False, normalize=False)
```

It **summarizes** the relationship between two or more categorical variables by counting the frequency of combinations. You can also apply an **`aggregation function`** like `sum`, `mean`, etc., to another variable.

| Parameter   | Description                                                                           |
| ----------- | ------------------------------------------------------------------------------------- |
| `index`     | Array-like or Series. Values to group by on the rows.                                 |
| `columns`   | Array-like or Series. Values to group by on the columns.                              |
| `values`    | (Optional) Array-like. If specified, computes aggregation using `aggfunc`.            |
| `aggfunc`   | Function to use for aggregation. Default is `count` (only when `values` is not None). |
| `margins`   | Adds row/column totals if `True`.                                                     |
| `normalize` | Normalize by row/column/all combinations. Helpful for proportions.                    |

**How `crosstab()` Is Useful?**

* **Analyzing categorical relationships**

* **Building confusion matrices**

* **Exploring frequency distributions**

* **Understanding interactions in surveys, demographics, or medical data**

* **Normalization to find proportions**

* It’s compact and readable compared to grouping manually.

### **Example 1: Basic Frequency Count:**

````python
import pandas as pd

df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'Smoker': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No']
})
print(df)
print("+"*50)
pd.crosstab(index= df['Gender'], columns= df['Smoker'])
# This tells us:
    # 2 female non-smokers, 1 female smoker
    # 1 male non-smoker, 2 male smokers
````

### **Example 2: Add Totals (`margins=True`):**

````python
df
````

````python
print(pd.crosstab(index= df['Gender'], columns= df['Smoker'], margins=True))
````

### **Example 3: Normalizing (Proportions):**

````python
# Normalize across the whole table
pd.crosstab(index= df['Gender'], columns= df['Smoker'], normalize=True)
#This shows proportions relative to the total.
````

````python
# Normalize by rows (per Gender)
pd.crosstab(index= df['Gender'], columns= df['Smoker'], normalize='index')
````

### **Example 4: Aggregation with `values` and `aggfunc`:**

Imagine we want to find average **Age** for combinations of **Gender** and **Smoker**.

````python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'Smoker': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No'],
    'Age': [25, 30, 22, 45, 28, 35]
})
df
````

````python
pd.crosstab(index= df['Gender'], columns= df['Smoker'], values=df['Age'], aggfunc='mean')
````

### **Example 5: Confusion Matrix (Classic Use Case):**

````python
actual = ['cat', 'dog', 'dog', 'cat', 'dog', 'cat']
predicted = ['cat', 'cat', 'dog', 'dog', 'dog', 'cat']
````

````python
actual
````

````python
predicted
````

````python
pd.crosstab(index= pd.Series(actual, name='Actual'),
            columns= pd.Series(predicted, name='Predicted'))
````

| Feature                     | Description                              |
| --------------------------- | ---------------------------------------- |
| Fast group-by on categories | Group and count frequencies quickly      |
| Supports aggregation        | Mean, sum, count on numeric variables    |
| Useful for comparison       | Analyze relationships between categories |
| Confusion matrix            | Evaluate classification models           |
| Optional normalization      | Proportion view (percentage style)       |
| Row/column totals           | Easy summary of group counts             |

## **When to Use:**

Use `pd.crosstab()` when:

* we need to **summarize relationships** between **two or more categorical variables**.
* we’re doing **exploratory data analysis (EDA)**.
* we’re analyzing **survey**, **health**, or **behavioral data**.
* we want a **confusion matrix** from model predictions.
