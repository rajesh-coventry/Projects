# **Adding Lines on Plots:**

In **Matplotlib**, we can add **lines** to plots to highlight specific positions (like thresholds, averages, or reference points). Two of the most useful functions for this are:

## **`plt.axvline()`** → **Add a vertical line**

## **`plt.axhline()`** → **Add a horizontal line**

```python
plt.axvline(x=position, color='color', linestyle='--', linewidth=1)
plt.axhline(y=position, color='color', linestyle='--', linewidth=1)
```

* **`x` or `y`:** Position of the line along the axis

* **`color`:** Line color (e.g., `'red'`, `'blue'`)

* **`linestyle`:** Dashed (`'--'`), dotted (`':'`), solid (`'-'`)

* **`linewidth`:** Thickness of the line

### **1. Vertical Line with `axvline()`:**

````python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Add vertical line at x = 2.5
plt.axvline(x=2.5, color='red', linestyle='--', linewidth=2)

plt.title('Vertical Line Example')
plt.show()
````

### **2. Horizontal Line with `axhline()`:**

````python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Add horizontal line at y = 22
plt.axhline(y=20, color='green', linestyle=':', linewidth=2)

plt.title('Horizontal Line Example')
plt.show()
````

### **3. Add Text to Lines:**

To **add text to lines** in plots, we can use the `plt.text()` function to place annotations near the lines.

````python
import matplotlib.pyplot as plt

# Plotting some sample data
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
plt.plot(x, y)

# Add a vertical line at x = 2.5
plt.axvline(x=2.5, color='red', linestyle='--')
plt.text(2.5, max(y), 'x = 2.5', rotation=90, color='red', ha='right', va='top')

# Add a horizontal line at y = 22
plt.axhline(y=22, color='green', linestyle=':')
plt.text(max(x), 22, 'y = 22', color='green', va='bottom', ha='right')

plt.title('Lines with Labels')
plt.show()
````

* `rotation=90`: Rotates the vertical line's label

* `ha='right'`, `va='top'`: Adjust horizontal and vertical alignment of text

* `max(x)` and `max(y)`: Used to position text near the plot edge

### **4. Using `annotate()` for More Flexibility:**

````python
# Plotting some sample data
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
plt.plot(x, y)

# Add a vertical line at x = 2.5
plt.axvline(x=2.5, color='red', linestyle='--')

# Add a horizontal line at y = 22
plt.axhline(y=22, color='green', linestyle=':')

plt.annotate('Threshold', xy=(2.5, 25), xytext=(3, 28),
             arrowprops=dict(arrowstyle='->', color='red'),
             color='red')

plt.title('Lines with Label')
plt.show()
````

**Tips:**

* You can combine both to highlight an intersection or threshold.

* These functions work well for annotating means, medians, or special conditions.

* They extend across the entire axis range (automatically scaled to plot limits).
