# **Set and Reset Index:**

The `set_index()` and `reset_index()` methods are fundamental tools for manipulating DataFrame indices. They work as complementary operations - one sets a new index while the other resets the index back to default.

#### **1. `set_index()` Method:**

The `set_index()` method converts one or more columns into the DataFrame's index (row labels). This is particularly useful when you want to use a specific column as the primary identifier for your rows.

```python
DataFrame.set_index(keys, drop=True, append=False, inplace=False, verify_integrity=False)
```

**Key Parameters:**
- `keys`: Column name(s) to set as index
- `drop`: Whether to delete the column(s) after setting as index (default: True)
- `append`: Whether to append to existing index (default: False)
- `inplace`: Whether to modify the DataFrame in place (default: False)

## **2. `reset_index()` Method:**

The `reset_index()` method converts the current index back to regular columns and creates a new default integer index.

**Syntax:**
```python
DataFrame.reset_index(level=None, drop=False, inplace=False, col_level=0, col_fill='')
```

**Key Parameters:**
- `level`: Which index levels to reset (for MultiIndex)
- `drop`: Whether to discard the index instead of converting to columns (default: False)
- `inplace`: Whether to modify the DataFrame in place (default: False)

````python
import pandas as pd
import numpy as np

# Example 1: Basic set_index() and reset_index()
print("=== Example 1: Basic Usage ===")
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
})

print("Original DataFrame:")
print(df)
print(f"Index: {df.index.tolist()}")
print()

# Set 'Name' as index
df_indexed = df.set_index('Name')
print("After set_index('Name'):")
print(df_indexed)
print(f"Index: {df_indexed.index.tolist()}")
print()

# Reset index back to default
df_reset = df_indexed.reset_index()
print("After reset_index():")
print(df_reset)
print(f"Index: {df_reset.index.tolist()}")
````

````python
# Example 2: Multiple columns as index (MultiIndex)
print("======== Example 2: MultiIndex ========")
df_sales = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Quarter': ['Q1', 'Q1', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Sales': [100, 150, 120, 180, 90, 160],
    'Profit': [20, 30, 25, 35, 18, 32]
})

print("Original sales DataFrame:")
print(df_sales)
print()

# Set multiple columns as index
df_multi = df_sales.set_index(['Region', 'Product'])
print("After set_index(['Region', 'Product']):")
print(df_multi)
print(f"Index levels: {df_multi.index.names}")
print()

# Reset only one level of MultiIndex
df_partial_reset = df_multi.reset_index(level='Product')
print("After reset_index(level='Product'):")
print(df_partial_reset)
print()

# Reset all levels
df_full_reset = df_multi.reset_index()
print("After reset_index() (all levels):")
print(df_full_reset)
````

````python
# Example 3: Using drop parameter
print("=== Example 3: Using drop parameter ===")
df_temp = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Price': [800, 25, 50, 300]
})

print("Original DataFrame:")
print(df_temp)
print()

# Set index without dropping the column
df_no_drop = df_temp.set_index('ID', drop=False)
print("set_index('ID', drop=False) - keeps ID column:")
print(df_no_drop)
print()

# Reset index and drop it
df_drop_reset = df_no_drop.reset_index(drop=True)
print("reset_index(drop=True) - discards the index:")
print(df_drop_reset)
````

````python
# Example 4: Using append parameter
print("=== Example 4: Using append parameter ===")
df_student = pd.DataFrame({
    'StudentID': [1, 2, 3, 4],
    'Name': ['John', 'Jane', 'Mike', 'Sarah'],
    'Grade': ['A', 'B', 'A', 'C'],
    'Subject': ['Math', 'Science', 'Math', 'Science']
})

print("Original DataFrame:")
print(df_student)
print()

# First set StudentID as index
df_with_id = df_student.set_index('StudentID')
print("After set_index('StudentID'):")
print(df_with_id)
print()

# Append Subject to existing index
df_appended = df_with_id.set_index('Subject', append=True)
print("After set_index('Subject', append=True):")
print(df_appended)
print(f"Index names: {df_appended.index.names}")
````

````python
# Example 5: Practical use case - Time series data
print("=== Example 5: Time Series Data ===")
dates = pd.date_range('2024-01-01', periods=6, freq='D')
df_timeseries = pd.DataFrame({
    'Date': dates,
    'Temperature': [20, 22, 19, 25, 23, 21],
    'Humidity': [65, 70, 68, 60, 72, 69],
    'Pressure': [1013, 1015, 1012, 1018, 1016, 1014]
})

print("Original time series DataFrame:")
print(df_timeseries)
print()

# Set Date as index for time series operations
df_ts_indexed = df_timeseries.set_index('Date')
print("After set_index('Date') - ready for time series operations:")
print(df_ts_indexed)
print()

# Now we can easily access data by date
print("Temperature on 2024-01-03:")
print(df_ts_indexed.loc['2024-01-03', 'Temperature'])
print()

# Reset for other operations
df_ts_reset = df_ts_indexed.reset_index()
print("After reset_index() - back to regular DataFrame:")
print(df_ts_reset)
````

````python
# Example 6: Working with duplicate indices
print("=== Example 6: Handling Duplicates ===")
df_dup = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'C', 'B'],
    'Value1': [10, 20, 15, 30, 25],
    'Value2': [100, 200, 150, 300, 250]
})

print("DataFrame with duplicate categories:")
print(df_dup)
print()

# Set Category as index (allows duplicates by default)
df_dup_indexed = df_dup.set_index('Category')
print("After set_index('Category'):")
print(df_dup_indexed)
print()

# Access all rows with Category 'A'
print("All rows with Category 'A':")
print(df_dup_indexed.loc['A'])
print()

# Group by index and calculate mean
print("Mean values by Category:")
print(df_dup_indexed.groupby(level=0).mean())
````

````python
# Example 7: Using inplace parameter
print("=== Example 7: Using inplace parameter ===")
df_inplace = pd.DataFrame({
    'Code': ['A1', 'B2', 'C3'],
    'Value': [100, 200, 300]
})

print("Original DataFrame (ID:", id(df_inplace), "):")
print(df_inplace)
print()

# Modify in place
print("Using set_index('Code', inplace=True):")
df_inplace.set_index('Code', inplace=True)
print("Same DataFrame object (ID:", id(df_inplace), "):")
print(df_inplace)
print()

# Reset in place
print("Using reset_index(inplace=True):")
df_inplace.reset_index(inplace=True)
print("Same DataFrame object (ID:", id(df_inplace), "):")
print(df_inplace)
````

````python
# Example 8: Performance comparison
print("=== Example 8: Common Use Cases ===")

# Creating sample data for analysis
np.random.seed(42)
df_analysis = pd.DataFrame({
    'Store': ['Store_A', 'Store_B', 'Store_C'] * 4,
    'Month': ['Jan', 'Jan', 'Jan', 'Feb', 'Feb', 'Feb', 
              'Mar', 'Mar', 'Mar', 'Apr', 'Apr', 'Apr'],
    'Sales': np.random.randint(1000, 5000, 12),
    'Expenses': np.random.randint(500, 2000, 12)
})

print("Sales data for analysis:")
print(df_analysis)
print()

# Set hierarchical index for easier grouping and analysis
df_hierarchical = df_analysis.set_index(['Store', 'Month'])
print("With hierarchical index (Store, Month):")
print(df_hierarchical)
print()

# Easy access to specific store data
print("Sales data for Store_A:")
print(df_hierarchical.loc['Store_A'])
print()

# Calculate profit and add as new column
df_hierarchical['Profit'] = df_hierarchical['Sales'] - df_hierarchical['Expenses']
print("With calculated Profit column:")
print(df_hierarchical)
print()

# Group by store and calculate totals
print("Total by Store:")
print(df_hierarchical.groupby(level='Store').sum())
print()

# Reset to use different analysis approach
df_flat = df_hierarchical.reset_index()
print("Back to flat structure for pivot operations:")
pivot_table = df_flat.pivot_table(values='Sales', index='Store', columns='Month', aggfunc='sum')
print("Pivot table of Sales by Store and Month:")
print(pivot_table)
````

#### **When to use set_index():**

1. **Time Series Analysis**: Set datetime columns as index for time-based operations and resampling

2. **Data Lookup**: Use unique identifiers as index for faster `.loc[]` access
3. **Grouping Operations**: Create hierarchical indices for complex groupby operations
4. **Data Alignment**: Pandas automatically aligns data based on index during operations
5. **Merge Operations**: Indices are used in join operations for better performance

#### **When to use reset_index():**

1. **Converting Index to Columns**: When you need the index values as regular columns for analysis

2. **Flattening MultiIndex**: Simplify complex hierarchical indices
3. **Preparing Data for Export**: Many export formats work better with simple integer indices
4. **After Group Operations**: GroupBy operations often create new indices that need resetting
5. **Data Cleaning**: Remove or reorganize problematic indices

### **Performance Considerations:**

- Setting appropriate indices can significantly speed up data access and filtering operations
- MultiIndex structures are powerful but can be memory-intensive for large datasets
- Use `inplace=True` to avoid creating copies of large DataFrames
- Consider using `drop=True` in `reset_index()` when you don't need the old index as a column

### **Best Practices:**

1. **Choose Meaningful Indices**: Use columns that you frequently filter or group by
2. **Avoid Duplicate Indices**: Unless specifically needed, duplicate indices can cause confusion
3. **Document Index Structure**: Clear naming helps maintain code readability
4. **Reset When Needed**: Don't keep complex indices if they're not being used
5. **Consider Memory Usage**: MultiIndex can increase memory consumption
