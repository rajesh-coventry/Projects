# **Plotting in Plotly:**

#### **1. General Outline for Creating Plots in Plotly (using `plotly.graph_objects`):**

This structure is similar to how you'd think in Matplotlib but follows Plotly's interactive logic.

```python
import plotly.graph_objects as go

# 1. Create a figure object
# This is the container for all plot elements: traces, layout, etc.
fig = go.Figure()

# 2. Add plotting elements (called "traces" in Plotly)
# A trace is a plot element like a line, bar, scatter point, etc.
fig.add_trace(go.Bar(
    x=['A', 'B', 'C'],           # X-axis values
    y=[10, 20, 15],              # Y-axis values
    name='Example Bar',          # Legend label
    marker_color='indianred'     # Bar color
))

# 3. Customize layout: title, axis labels, etc.
fig.update_layout(
    title={                             # Title settings
        'text': "Example Plot",
        'x': 0.5,                        # Center the title
        'xanchor': 'center',
        'font': dict(size=20)
    },
    xaxis=dict(                         # X-axis settings
        title='X Axis Label',
        tickangle=0,
        showgrid=True
    ),
    yaxis=dict(                         # Y-axis settings
        title='Y Axis Label',
        showgrid=True
    ),
    legend=dict(                        # Legend settings
        title='Legend',
        orientation='h',
        yanchor='bottom',
        y=1.02,
        xanchor='right',
        x=1
    ),
    template='plotly_white',            # Theme (you can try 'plotly_dark' etc.)
    height=400,                         # Figure height
    width=1150                           # Figure width
)

# 4. (Optional) Customize annotations, shapes, or interactivity
# Example: Add a horizontal line
fig.add_shape(
    type="line",
    x0=0,
    x1=1,
    y0=15,
    y1=15,
    xref='paper',
    yref='y',
    line=dict(color="RoyalBlue", width=2, dash="dash")
)

# 5. Show the figure
fig.show()
```

| Step | What          | Plotly Equivalent                                |
| ---- | ------------- | ------------------------------------------------ |
| 1    | Create figure | `go.Figure()`                                    |
| 2    | Add plots     | `fig.add_trace(go.Bar(...))`, `go.Scatter`, etc. |
| 3    | Customize     | `fig.update_layout(...)`                         |
| 4    | Enhance       | `fig.add_shape(...)`, `fig.add_annotation(...)`  |
| 5    | Display       | `fig.show()`                                     |

#### **2. `plotly.express`:**

Here’s a **general outline for creating plots using `plotly.express`**, the higher-level API in Plotly. It’s faster for common tasks like bar plots, line charts, and scatter plots, and it's perfect when you're working with a `DataFrame`.

```python
import plotly.express as px
import pandas as pd

# 1. Prepare data (ideally in a pandas DataFrame)
# This makes it easier to work with plotly.express
data = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'A', 'B', 'C'],
    'Subgroup': ['X', 'X', 'X', 'Y', 'Y', 'Y'],
    'Value': [10, 15, 13, 17, 20, 14]
})

# 2. Create a chart using a px.<chart_type>() function
# Available types: bar, line, scatter, pie, box, etc.
fig = px.bar(
    data_frame=data,
    x='Category',            # X-axis column name
    y='Value',               # Y-axis column name
    color='Subgroup',        # Hue/grouping
    barmode='group',         # 'group', 'stack', etc.
    text='Value',            # Show values on top of bars
    title='Grouped Bar Chart Example:'  # Title of the plot
)

# 3. Update axis labels, ticks, layout
fig.update_layout(
    xaxis_title='Category Label',
    yaxis_title='Value Label',
    title_font_size=20,
    title_font_weight= "bold", 
    title_x=0.5,  # Center the title
    legend_title_text='Subgroup',
    template='plotly_white',  # Clean background theme
    height=400,
    width=1150
)

# 4. (Optional) Fine-tune appearance
fig.update_traces(
    textposition='outside',     # Show text outside bars
    marker_line_width=1.5,      # Border around bars
    marker_line_color='black'
)

# 5. Show the figure
fig.show()
```

| Step | What You Do      | How in Plotly Express                   |
| ---- | ---------------- | --------------------------------------- |
| 1    | Prepare data     | Use `pandas.DataFrame`                  |
| 2    | Plot             | `px.bar()`, `px.line()`, `px.scatter()` |
| 3    | Customize layout | `fig.update_layout()`                   |
| 4    | Enhance traces   | `fig.update_traces()`                   |
| 5    | Show             | `fig.show()`                            |

#### **Common Plotly Express Functions:**
