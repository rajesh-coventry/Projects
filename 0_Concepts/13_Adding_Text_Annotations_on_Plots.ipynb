# **Adding Text Annotations on the Plots:**

> **`plt.text()`**

The `plt.text()` function in Matplotlib is used to **add text annotations** at specific coordinates on a plot.

```python
plt.text(x, y, s, fontdict=None, **kwargs)
```

* `x`, `y`: Coordinates where the text should appear.
* `s`: The string (text) to display.
* `fontdict` (optional): Dictionary to set font properties.
* `**kwargs`: Additional formatting like `fontsize`, `color`, `rotation`, etc.

**Common Uses:**

* Label specific data points
* Add values above bars in a bar plot
* Annotate graphs with information
* Add labels dynamically using `enumerate()`

**Example- 1:**

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [10, 20, 15]

plt.plot(x, y, marker='o')

# Add text at each point
for i in range(len(x)):
    plt.text(x= x[i], y= y[i], s= f'({x[i]},{y[i]})', fontsize= 15, color='red')

plt.show()
````

**Example-2:**

````python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# Define font properties
font_settings = {
    'family': 'serif',
    'color':  'darkred',
    'weight': 'bold',
    'size': 12,
}

# Plot the data
plt.plot(x, y, marker='o', linestyle='-')

# Annotate a specific point
plt.text(x= x[1], y= y[1] + 1, s= "Peak", fontdict=font_settings)

plt.title('Line Plot with Custom Text')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

**Example-3:**

````python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# Define font properties
font_settings = {
    'family': 'serif',
    'color':  'darkred',
    'weight': 'bold',
    'size': 12,
}

# Plot the data
plt.plot(x, y, marker='o', linestyle='-')

for i in range(len(x)):
    plt.text(x= x[i]+0.1, y= y[i]+0.5, s= f'({x[i]},{y[i]})', color='red', fontdict=font_settings)

plt.title('Line Plot with Custom Text')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

**Example-4:** **Using `enumerate()` with `plt.text()`:**

This is useful when you want to access **both index and value** in a loop.

````python
import matplotlib.pyplot as plt

values = [25, 40, 35, 50]
labels = ['A', 'B', 'C', 'D']

plt.bar(labels, values)

for i, val in enumerate(values):
    plt.text(x= i, y= val + 1, s= str(val), ha='center', fontsize=10, color='black')

plt.ylim(0, 60)
plt.show()
````

* `i` is the index (used as x-position),
* `val` is the bar height (used as y-position),
* `val + 1` offsets the text slightly above the bar,
* `ha='center'` centers the text horizontally over the bar.

### **Tips:**

* Use `ha='center'`, `va='bottom'` for proper alignment.
* Offset text slightly so it doesn’t overlap data points.
* Combine with conditionals (e.g., highlight values above a threshold).
