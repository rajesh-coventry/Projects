# **Stacking:**

## **NumPy Stacking:**

#### **1. Horizontal Stacking (Side-by-side / column-wise):**

> `np.hstack()`    
> 
> or 
> 
> `np.concatenate(..., axis=1)`
> 

* Joins arrays **`along columns`**.
* Arrays must have the **`same number of rows`** (same shape along `axis=0`).

```python
import numpy as np

a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])
```

```python
a
```

```python
b
```

```python
# Horizontal stacking
np.hstack((a, b))
```

```python
np.concatenate((a, b), axis=1)
```

##### **2. Vertical Stacking (Top-to-bottom / row-wise):**

> `np.vstack()` 
> 
> or 
> 
> `np.concatenate(..., axis=0)`

* Joins arrays **`along rows`**.
* Arrays must have the **`same number of columns`** (same shape along `axis=1`).

```python
# Vertical stacking
np.vstack((a, b))
```

```python
# or
np.concatenate((a, b), axis=0)
```

## **Pandas Stacking:**

In pandas, we typically work with **`Series`** and **`DataFrames`**.

##### **1. Horizontal Stacking in pandas:**

> `pd.concat([...], axis=1)`

Also: 
> `DataFrame.join()` or 
> 
> `DataFrame.merge()` (for key-based joins)

* Adds **`columns`** side-by-side.
* Useful when combining features or multiple datasets with the same index.

```python
import pandas as pd

s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])
```

```python
s1
```

```python
s2
```

```python
# Horizontal stacking
pd.concat([s1, s2], axis=1)
```

##### **2. Vertical Stacking in pandas:**

> `pd.concat([...], axis=0)`

* Adds **`rows`** on top of each other.
* Useful when stacking data with the same columns.

```python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
```

```python
df1
```

```python
df2
```

```python
# Vertical stacking
pd.concat([df1, df2], axis=0)
```

> Use `ignore_index=True` if you want to reset row numbers.

| Stacking Type  | NumPy Function                             | pandas Equivalent          |
| -------------- | ------------------------------------------ | -------------------------- |
| **Horizontal** | `np.hstack()` / `concatenate(..., axis=1)` | `pd.concat([...], axis=1)` |
| **Vertical**   | `np.vstack()` / `concatenate(..., axis=0)` | `pd.concat([...], axis=0)` |
