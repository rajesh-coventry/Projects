# **Dummies:**

**One-hot encoding** is a method used to represent **categorical variables** as **binary vectors** (0s and 1s). It's commonly used in machine learning and deep learning when working with data that contains non-numeric categories.

Most ML models (like logistic regression, neural networks, SVMs) **don't work well with categorical data in text format** — they need numbers. But simply converting categories to numbers (like `Red = 1`, `Green = 2`, `Blue = 3`) can be misleading because it introduces **false ordinal relationships**.

One-hot encoding avoids this by giving **each category its own binary column**.

> One-hot encoding gives **`each category its own binary column`**.

### **Example:**

Suppose we have this column:

```python
Color
----
Red
Green
Blue
Red
```

Using **one-hot encoding**, this becomes:

| Color\_Blue | Color\_Green | Color\_Red |
| ----------- | ------------ | ---------- |
| 0           | 0            | 1          |
| 0           | 1            | 0          |
| 1           | 0            | 0          |
| 0           | 0            | 1          |

Each row has exactly **one `1`** and the rest are `0`.

#### **In pandas:**

````python
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
one_hot = pd.get_dummies(df)
one_hot
````

**✅ Pros:**

* No assumption of order between categories.
* Works well with most machine learning models.

**❌ Cons:**

* Increases dimensionality (especially with many unique values).
* Can lead to sparse data (many 0s), which isn't ideal for all algorithms.

### **Related Techniques:**

| Technique        | Description                                                             |
| ---------------- | ----------------------------------------------------------------------- |
| `One-hot encoding` | Binary column for each category; one active per row                     |
| `Label encoding`   | Each category gets an integer (useful for tree-based models)            |
| `Binary encoding`  | Categories encoded as binary strings (lower dimensional than one-hot)   |
| `Target encoding`  | Categories replaced with mean target values (used in some ML pipelines) |

In **pandas**, the concept of **"`dummies`"** refers to converting **`categorical variables`** into a format that can be provided to machine learning algorithms — specifically, creating **`binary (0 or 1) columns`** for each category level. This process is commonly called **`one-hot encoding`**.

**Suppose we have a column with categories like this:**

````python
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

**To convert this into a machine-learning-friendly format:**

````python
pd.get_dummies(df)
````

This means each category becomes a separate column, and only one column is `1` for each row — indicating the presence of that category.

#### **When is this concept useful?**

Dummies are useful when:

1. **`Feeding categorical data into machine learning models`** (e.g., linear regression, logistic regression, tree-based models).

2. **`Avoiding implicit ordering`** of categorical values (which happens if you use label encoding).

3. **`Ensuring models treat categories as distinct`** (not numerical quantities).

### 1. **`pd.get_dummies()`:**

Main function in pandas to create dummies.

```python
pd.get_dummies(data, columns=None, prefix=None, drop_first=False, dtype='uint8')
```

* `columns`: Specify which columns to encode.
* `drop_first=True`: Drops the first category to avoid multicollinearity (used in linear models).
* `prefix`: Add custom prefix to column names.
* `dtype`: Set output data type.

````python
pd.get_dummies(df, drop_first=True)
````

This drops one dummy column, which helps in regression models by avoiding dummy variable trap.

#### 2. **`pd.from_dummies()`:**

This **`reverses`** the effect of `get_dummies()` — it reconstructs the original categorical column from dummies.

````python
dummies = pd.get_dummies(df)
pd.from_dummies(dummies)
````

This is helpful for **`converting predictions or transformed data back to the original categories`**.

#### 3. **`Series.str.get_dummies()`:**

We can call `get_dummies()` **directly on a string Series**. It's useful when you have **multiple labels in a single string**, separated by a delimiter (like in multi-label problems).

````python
s = pd.Series(['a|b', 'a|c', 'b|c'])
s
````

````python
s.str.get_dummies(sep='|')
````

## **Different Encoding Methods:**

### **1. One-Hot Encoding:**

- Converts categorical values into binary columns (one per category).

- **pandas:** `pd.get_dummies(df)`
- **sklearn:** `OneHotEncoder()` from `sklearn.preprocessing`

````python
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
# Using pandas
one_hot = pd.get_dummies(df)
print(one_hot)
````

### **2. Label Encoding:**

- Assigns an integer to each category (useful for tree-based models, not for linear models).

* **sklearn:** `LabelEncoder()` from `sklearn.preprocessing`
  *(Only works on 1D arrays/Series)*

````python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
le = LabelEncoder()
df['Color_encoded'] = le.fit_transform(df['Color'])
df
````

### **3. Ordinal Encoding:**

- Encodes categories as integers **with assumed order** (e.g., low < medium < high).

- **sklearn:** `OrdinalEncoder()` from `sklearn.preprocessing`
  *(Use `categories=` argument to define order)*

````python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

df = pd.DataFrame({'Size': ['Small', 'Medium', 'Large', 'Small']})
df
````

````python
encoder = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])
df['Size_encoded'] = encoder.fit_transform(df[['Size']])
df
````

### **4. Binary Encoding:**

* Converts categories to binary numbers and splits each bit into separate columns.
*(Reduces dimensionality compared to one-hot.)*

* **library:** `BinaryEncoder()` from `category_encoders` package

````python
import pandas as pd
from category_encoders import BinaryEncoder

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
be = BinaryEncoder()
df_encoded = be.fit_transform(df)
df_encoded
````

### **5. Target Encoding (Mean Encoding):**

* Replaces each category with the **mean of the target variable** for that category.

* **library:** `TargetEncoder()` from `category_encoders`

````python
import pandas as pd
from category_encoders import TargetEncoder

df = pd.DataFrame({
    'Color': ['Red', 'Green', 'Blue', 'Red'],
    'Target': [1, 0, 1, 1]
})
df
````

````python
te = TargetEncoder()
df['Color_encoded'] = te.fit_transform(df['Color'], df['Target'])
df
````

### **6. Frequency Encoding:**

* Replaces each category with its **frequency count or proportion**.

* **manual (pandas):**

  ```python
  df['col'] = df['col'].map(df['col'].value_counts())
  ```

````python
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
freq = df['Color'].value_counts()
df['Color_encoded'] = df['Color'].map(freq)
df
````

### **7. Hash Encoding:**

* Hashes categories into a fixed number of columns; useful for high-cardinality data.

* **library:** `HashingEncoder()` from `category_encoders`

````python
import pandas as pd
from category_encoders import HashingEncoder

df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
df
````

````python
he = HashingEncoder(n_components=4)  # Number of hash columns
df_encoded = he.fit_transform(df)
df_encoded
````
