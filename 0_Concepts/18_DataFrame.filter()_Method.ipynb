# **Lable based Filtering:**

The `.filter()` method is used to **select rows or columns** from a DataFrame based on **labels (names)** — not values. It's a **`label-based`** filter, unlike `.iloc` (which is index-based) or `.loc` (which is label + condition based).

```python
> DataFrame.filter(items=None, like=None, regex=None, axis=None)
```

* **`items`** → List of exact labels (e.g., column or row names).
* **`like`** → Substring to match in labels.
* **`regex`** → Regular expression pattern to match labels.
* **`axis`** → `0` (rows) or `1` (columns); defaults to `0`.

**Example Setup:**

````python
import pandas as pd

data = {
    "Customer_ID": [1, 2, 3],
    "Customer_Name": ["Alice", "Bob", "Charlie"],
    "Customer_Age": [25, 30, 35],
    "Product_Purchased": ["Laptop", "Tablet", "Phone"],
    "Purchase_Amount": [1200, 800, 600]
}

df = pd.DataFrame(data)
df
````

**`Example 1`: Select specific columns using `items`:**

````python
# Select exact columns by name
df.filter(items=["Customer_Name", "Purchase_Amount"], axis=1)
````

This returns a new DataFrame with only `Customer_Name` and `Purchase_Amount`.

**`Example 2`: Select columns containing a substring using `like`:**

````python
# Select columns whose names contain "Customer"
df.filter(like="Customer", axis=1)
````

**`Example 3`: Select columns using `regex`:**

````python
# Select columns starting with "Product"
df.filter(regex="^Product", axis=1)
````

**`Example 4`: Filter rows by index label (less common, but possible):**

````python
# Set index first
df_with_index = df.set_index("Customer_ID")
df_with_index
````

````python
# Select specific rows by index label
df_with_index.filter(items=[1, 3], axis=0)
````

#### **When to use `.filter()`?**

* When you want to **match multiple columns based on patterns** (e.g., all columns starting with "Customer").

* When you want a **cleaner alternative to `df.loc[:, ...]` or `df[[...]]`** for label-based filtering.

* When working with **high-dimensional data** (e.g., many columns or hierarchical index).


| Use Case            | Recommended Method               |
| ------------------- | -------------------------------- |
| Exact column names  | `df.filter(items=[...], axis=1)` |
| Substring in names  | `df.filter(like="...", axis=1)`  |
| Regex pattern match | `df.filter(regex="...", axis=1)` |
| Filter by row index | `df.filter(..., axis=0)`                |

## **`DataFrame.set_index()` Method:**

The `dataframe.set_index()` method is used to set one or more columns of a DataFrame as its index. This can enhance the organization of data and improve the efficiency of data retrieval and manipulation.

**`Data Alignment`**:   
When performing operations between DataFrames, aligned indices can help ensure correct alignment of data.

**`Hierarchical Indexing`**:  
You can create a multi-level index (hierarchical index) for more complex data structures.

##### **1. Simple Indexing:**  
Suppose we have a DataFrame with sales data, and we want to set the "Date" column as the index:

````python
import pandas as pd

data = {
    'Date': ['2023-01-01', '2023-01-02', '2023-01-03'],
    'Sales': [200, 250, 300]
}

df = pd.DataFrame(data)
df.set_index('Date', inplace=True)
df
````

##### **2. Multi-level Indexing:**   
We might want to analyze sales data by region and product. 

We can create a hierarchical index:

````python
data = {
    'Region': ['North', 'North', 'South', 'South'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [200, 150, 300, 250]
}

df = pd.DataFrame(data)
df
````

````python
df.set_index(['Region', 'Product'], inplace=True)
df
````

##### **3. Resetting the Index:**
If we need to revert back to the default integer index, we can use `reset_index()`:

````python
df
````

````python
df.reset_index(inplace=True)
df
````

##### **4. Data Alignment During Merge:**
If we are merging two DataFrames, having a properly set index can help ensure that the data aligns correctly:

````python
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value2': [4, 5, 6]})
df1
````

````python
df2
````

````python
df1.set_index('Key', inplace=True)
df2.set_index('Key', inplace=True)

merged_df = df1.join(df2)
merged_df
````

The `set_index()` method is a powerful tool in pandas that allows for better organization of data and facilitates complex data operations. By setting meaningful indices, you can improve both the readability and performance of your DataFrame manipulations.
