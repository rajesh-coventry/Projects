# **Dataframe.rename() Method:**

The `DataFrame.rename()` method is used to rename labels (`columns` or `index`) in a DataFrame. 

> * **Columns** using the `columns` parameter.

> * **Index (rows)** using the `index` parameter.

It returns a new DataFrame (unless `inplace=True` is set), with the updated names.

```python
DataFrame.rename(mapper=None, *, index=None, columns=None, axis=None, inplace=False, errors='ignore')
```

* `index`: dict-like or function to rename row labels.

* `columns`: dict-like or function to rename column labels.

* `inplace`: If True, modifies the object directly.

* `errors`: If `'raise'`, raises an error for invalid labels; if `'ignore'`, skips them.

While pandas itself doesn’t have many built-in datasets, `seaborn` (which uses pandas) does. Let's use Seaborn's `tips` dataset via `pandas`.

````python
# Load a sample dataset from seaborn (hosted online)
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv')

print(df.head())
````

````python
df.shape
````

````python
df.columns
````

**Rename columns:**

````python
# Rename 'total_bill' to 'bill_amount', and 'tip' to 'gratuity'
renamed_df = df.rename(columns={'total_bill': 'bill_amount', 'tip': 'gratuity'})
print(renamed_df.head())
````

**Rename rows (index):**

````python
df.head(5)
````

````python
# Rename first 3 rows
renamed_df = df.rename(index={0: 'first', 1: 'second', 2: 'third'})
print(renamed_df.head())
````

**Using a function to rename column names (e.g., uppercase):**

````python
# Convert all column names to uppercase: 
renamed_df = df.rename(columns=str.upper)
print(renamed_df.columns) 
````

**`inplace=True` Example:**

````python
df.rename(columns={'sex': 'gender'}, inplace=True)
print(df.columns)
````
