# **Filtering:**

**1. Filtering Based on Column Names or Index Values:**

| Function / Method  | One-line Description                            |
| ------------------ | ----------------------------------------------- |
| `df.columns`       | Returns a list of column names.                 |
| `df.index`         | Returns the row index (labels).                 |
| `df.filter()`      | Filters columns or index using `labels` or `regex`. |
| `df.loc[]`         | Access rows/columns by `label-based` indexing.    |
| `df.iloc[]`        | Access rows/columns by `position-based` indexing. |
| `df.drop()`        | Removes rows or columns by `labels`.              |
| `df.set_index()`   | Sets one or more columns as the `index`.          |
| `df.reset_index()` | Resets index to `default integer index`.          |
| `df.reindex()`     | Changes the row and/or column labels.           |

**2. Filtering Based on Actual Values in the DataFrame:**

| Function / Method                    | One-line Description                                                        |
| ------------------------------------ | --------------------------------------------------------------------------- |
| `df[df['col'] == value]`             | Filters rows where column equals a value.                                   |
| `df[df['col'] > value]`              | Filters rows where column is greater than a value.                          |
| `df[df['col'].isin([v1, v2])]`       | Filters rows where column value is in a list.                               |
| `df[df['col'].str.contains('text')]` | Filters rows where string column contains a substring.                      |
| `df.query('col > 10')`               | Filters rows using a query string.                                          |
| `df.where(condition)`                | Returns the same shape with rows/columns not matching the condition as `NaN`. |
| `df.mask(condition)`                 | Opposite of `where()`, masks values where condition is True.                |
| `df.dropna()`                        | Drops rows or columns with `NaN` values.                                      |
| `df.notna()` / `df.isna()`           | Boolean mask for non-null/null values.                                      |
| `df.duplicated()`                    | Flags duplicate rows.                                                       |
| `df.nunique()`                       | Returns number of unique values per column.                                 |
| `df.between(a, b)`                   | Filters values between two numbers.                                         |
| `df.any()` / `df.all()`              | Checks if any/all values meet a condition across axis.                      |
