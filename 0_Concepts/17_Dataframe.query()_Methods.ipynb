# **Filtering `Dataframe.query()` Method:**

The `DataFrame.query()` method in pandas allows you to filter DataFrames using a string expression that resembles natural language, making your code more readable and intuitive compared to traditional boolean indexing.

> `DataFrame.query(expr, inplace=False, **kwargs)`

The `expr` parameter is a string containing the query expression that evaluates to a boolean condition.

### **When It's Useful:**

**The `query()` method is particularly valuable when:**

- You have complex filtering conditions that would be verbose with traditional boolean indexing

- Column names contain spaces or special characters

- You want more readable code that resembles SQL-like syntax

- You need to combine multiple conditions with logical operators

- You're working with large datasets where query optimization matters

#### **1. Basic Filtering:**

```python
import pandas as pd 

# Sample data
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'salary': [50000, 60000, 70000, 55000, 65000],
    'department': ['HR', 'IT', 'Finance', 'HR', 'IT']
})
df
```

```python
# Traditional way 
result1 = df[df['age'] > 30]

# Using query() - more readable
result2 = df.query('age > 30')
```

```python
result1
```

```python
result2
```

#### **2. Multiple Conditions:**

```python
df
```

```python
# Traditional way - can get messy
result1 = df[(df['age'] > 25) & (df['salary'] < 65000)]

# Using query() - cleaner
result2 = df.query('age > 25 and salary < 65000')
```

```python
result1
```

```python
result2
```

#### **3. Working with Column Names Containing Spaces:**

```python
# DataFrame with spaces in column names
df_spaces = pd.DataFrame({
    'first name': ['Alice', 'Bob', 'Charlie'],
    'last name': ['Smith', 'Jones', 'Brown'],
    'annual salary': [50000, 60000, 70000]
})
df_spaces
```

```python
# Traditional way requires backticks or brackets
result1 = df_spaces[df_spaces['annual salary'] > 55000]

# Using query() with backticks for spaced names
result2 = df_spaces.query('`annual salary` > 55000')
```

```python
result1
```

```python
result2
```

### **4. String Operations:**

```python
df
```

```python
# Filter by string conditions
result_a = df.query('department == "IT"')

# String contains
result = df.query('name.str.contains("a", case=False)')
```

```python
result_a
```

```python
result
```

#### **5. Using Variables in Queries:**

```python
df
```

```python
min_age = 28
target_dept = 'HR'

# Use @ to reference external variables
result = df.query('age >= @min_age and department == @target_dept')
result
```

#### **6. Complex Expressions:**

```python
df
```

```python
# Multiple conditions with different operators
result_b = df.query('age > 25 and (department == "IT" or salary > 60000)')

# Using isin() equivalent
departments = ['IT', 'Finance']
result_c = df.query('department in @departments')
```

```python
result_b
```

```python
result_c
```

#### **7. Index-based Queries:**

```python
df
```

```python
# Set age as index
df_indexed = df.set_index('age')

# Query on index
result_z = df_indexed.query('age > 28 and age < 35')
```

```python
df_indexed
```

```python
result_z
```

#### **8. Comparison with Traditional Methods:**

```python
# Traditional boolean indexing
mask = (df['age'] > 25) & (df['salary'] < 65000) & (df['department'].isin(['IT', 'HR']))
traditional_result = df[mask]

# Equivalent query() - more readable
query_result = df.query('age > 25 and salary < 65000 and department in ["IT", "HR"]')
```

```python
traditional_result
```

```python
query_result
```

The `query()` method can be faster for large DataFrames because it uses numexpr under the hood for optimization when available. However, for small DataFrames, traditional boolean indexing might be slightly faster due to less overhead.

### **Limitations:**

   - String expressions need to be valid Python expressions
   - Some complex operations might still require traditional indexing
   - Column names with special characters need backticks
   - Less IDE support for syntax highlighting and autocomplete compared to regular Python code
