# **Aggregation:**

In **pandas**, `.agg()` and `.aggregate()` are **used to apply one or more aggregation functions** (like `sum()`, `mean()`, `min()`, etc.) on a `DataFrame` or `Series`. They are **aliases of each other**, so `.agg()` and `.aggregate()` behave the same.

**They’re commonly used with:**

* `groupby()` for grouped aggregation.
* `DataFrame`/`Series` directly for column-wise or row-wise aggregation.

```python
DataFrame.agg(func, axis=0, *args, **kwargs)
```

* `func`: A single function, list of functions, or a dictionary of column-function mappings.
* `axis=0`: Aggregation is applied column-wise (default).
* `axis=1`: Aggregation is applied row-wise.

### ▶ **1. Aggregating a DataFrame using a single function:**

````python
import pandas as pd

# Sample DataFrame
data = {
    'Math': [90, 80, 70, 60],
    'Science': [88, 76, 95, 89],
    'English': [78, 85, 88, 92]
}

df = pd.DataFrame(data)
print(df)
print("-"*25)

# Apply a single aggregation function: mean
mean_values = df.agg('mean')  # or df.aggregate('mean')
print("Column-wise mean:")
print(mean_values)
````

This calculates the **mean of each column**.

### ▶ **2. Aggregating with multiple functions:**

````python
df
````

````python
# Apply multiple aggregation functions to each column
multi_agg = df.agg(['mean', 'min', 'max'])
print("Multiple aggregations:")
multi_agg
````

This gives `mean`, `min`, and `max` for each column.

````python
df
````

````python
# Apply multiple aggregation functions for  each row for numeric values only:
multi_agg = df[["Math", "Science", "English"]].agg(['mean', 'min', 'max'], axis= 1)
print("Multiple aggregations:")
multi_agg
````

### ▶ **3. Aggregating different functions per column using a dictionary:**

````python
df
````

````python
# Custom aggregation for each column
custom_agg = df.agg({
    'Math': ['mean', 'std'],
    'Science': ['min', 'max'],
    'English': 'sum'
})
print("Custom aggregation per column:")
print(custom_agg)
````

Dictionary syntax lets you **specify different aggregations per column**.

### ▶ **4. Row-wise aggregation (`axis=1`):**

````python
df
````

````python
# Aggregate across rows instead of columns
row_wise_sum = df.agg('sum', axis=1)
print("Row-wise sum:")
row_wise_sum
````

This gives **total marks per student** (row-wise sum).

### ▶ **5. With `groupby()`:**

````python
# Sample DataFrame with groups
data = {
    'Student': ['Alice', 'Bob', 'Alice', 'Bob'],
    'Subject': ['Math', 'Math', 'Science', 'Science'],
    'Score': [90, 80, 85, 75]
}
df2 = pd.DataFrame(data)
df2
````

````python
# Group by student and aggregate scores
grouped = df2.groupby('Student')['Score'].agg(['mean', 'max'])
print("\nGroupby aggregation:")
grouped
````

Useful when you want to **summarize grouped data**.

## **`agg()` with user-defined functions (UDFs):**

````python
import pandas as pd
import numpy as np

# Sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math': [90, 80, 70, 60],
    'Science': [88, 76, 95, 89],
    'English': [78, 85, 88, 92]
}

df = pd.DataFrame(data)
df
````

### **1. Using a custom function defined with `def`:**

````python
# Custom function: range of values
def value_range(series):
    return series.max() - series.min()

# Apply custom function to all columns
result = df[['Math', 'Science', 'English']].agg(value_range)
print("Value range (max - min) for each subject:")
print(result)
````

### **2. Using `lambda` functions:**

````python
# Use a lambda function directly
result = df[['Math', 'Science', 'English']].agg(lambda x: x.mean() + x.std())
print("\nMean + Std Dev for each subject:")
print(result)
````

### **3. Using multiple custom functions:**

````python
# Define custom functions
def custom_stat1(x):
    return x.mean() * 2

def custom_stat2(x):
    return np.median(x)

# Apply multiple custom functions
result = df[['Math', 'Science']].agg([custom_stat1, custom_stat2])
print("\nMultiple custom functions applied to Math and Science:")
print(result)
````

### **4. Using a dictionary to apply different custom functions to different columns:**

````python
# Define functions
def range_fn(x):
    return x.max() - x.min()

def mean_plus_std(x):
    return x.mean() + x.std()

# Use dictionary to map columns to custom functions
result = df.agg({
    'Math': range_fn,
    'Science': mean_plus_std,
    'English': lambda x: x.sum() / len(x)  # same as mean
})
print("\nDifferent custom functions per column:")
print(result)
````

### **5. Custom aggregation with `groupby()`:**

````python
import warnings
warnings.filterwarnings("ignore")
# Group by the first letter of name
df['Group'] = df['Name'].str[0]  # A, B, C, D

# Group by 'Group' and aggregate scores
grouped = df.groupby('Group', observed=True)[['Math', 'Science', 'English']].agg({
    'Math': lambda x: x.max() - x.min(),  # range
    'Science': np.mean,                   # mean
    'English': np.std                     # standard deviation
})
print("\nGroupby with custom aggregation functions:")
print(grouped)
````

### **6. Returning a named custom metric:**

````python
# Custom function that returns a named value
def score_summary(x):
    return pd.Series({
        'range': x.max() - x.min(),
        'custom_avg': (x.mean() + x.median()) / 2
    })

# Apply to a column
summary = df['Math'].agg(score_summary)
print("\nNamed summary statistics for Math:")
print(summary)
````

| Type             | Example                            |
| ---------------- | ---------------------------------- |
| Single function  | `df.agg(my_func)`                  |
| Lambda function  | `df.agg(lambda x: x.mean())`       |
| Multiple funcs   | `df.agg([func1, func2])`           |
| Dict mapping     | `df.agg({'col1': f1, 'col2': f2})` |
| With `groupby()` | `df.groupby(...).agg(...)`         |
