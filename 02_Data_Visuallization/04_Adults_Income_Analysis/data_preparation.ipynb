| Column             | Meaning                                                                                                                                                                   |
| ------------------ | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **age**            | Age of the individual (continuous variable).                                                                                                                              |
| **workclass**      | Type of employment or class of work (e.g., 'Private', 'Self-emp', 'Government', etc.). Represents socio-economic status.                                                  |
| **fnlwgt**         | Final weight (used to adjust for survey sampling design). It indicates how many people the census believes a person represents. Often not useful for predictive modeling. |
| **education**      | Highest level of education completed (e.g., 'Bachelors', 'HS-grad', 'Masters', etc.).                                                                                     |
| **education-num**  | Numerical representation of education level (e.g., 'Bachelors' = 13). More machine-readable than `education`.                                                             |
| **marital-status** | Marital status (e.g., 'Married-civ-spouse', 'Divorced', 'Never-married', etc.).                                                                                           |
| **occupation**     | Job type or field (e.g., 'Tech-support', 'Craft-repair', 'Sales', etc.).                                                                                                  |
| **relationship**   | Relationship of the individual to others in the household (e.g., 'Husband', 'Not-in-family', 'Own-child', etc.).                                                          |
| **race**           | Race of the individual (e.g., 'White', 'Black', 'Asian-Pac-Islander', etc.).                                                                                              |
| **gender**         | Gender of the individual ('Male' or 'Female').                                                                                                                            |
| **capital-gain**   | Income from capital gains (e.g., profits from selling stocks or property). Mostly zeros.                                                                                  |
| **capital-loss**   | Losses from capital assets (e.g., losses in investments). Also mostly zeros.                                                                                              |
| **hours-per-week** | Number of hours worked per week (continuous variable).                                                                                                                    |
| **native-country** | Country of origin (e.g., 'United-States', 'Mexico', 'Philippines', etc.).                                                                                                 |
| **salary**         | Target variable: income class. Either '`<=50K`' or '`>50K`', indicating whether the individual earns more or less than \$50,000 annually.                                 |

**Age:**

1. The total number of observations (rows or data records) in the dataset is **48,842**.

2. The **average (mean) age** of individuals in the dataset is approximately **38.6 years**.

3. The **standard deviation (std)** of **13.7 years** means that **most ages (about 68% under   normal distribution) fall within 13.7 years above or below the mean**.
   So, roughly 68% of people have an age between **(38.6 - 13.7 = 24.9)** and **(38.6 + 13.7 = 52.3)**. It's not a guarantee for a *random person*, but rather a description of the **spread or variability** in the data. If age is not normally distributed, then the interpretation of the standard deviation as a simple `“plus or minus range around the mean”`  does not reliably hold true. In such cases we use `percentiles`, `boxplots`, `histograms` and `density plots` to approximate the distribution of variables (=age in this example). 

4. The **minimum age** recorded is **17 years**.

5. The **maximum age** recorded is **90 years**.

6. The **25th percentile (Q1)** being **28 years** means **25% of the people are aged 28 or younger**.

7. The **50th percentile (median)** being **37 years** means **50% of the people are aged 37 or younger**.

8. The **75th percentile (Q3)** being **48 years** means **75% of the people are aged 48 or younger**.

**Capital Gain:**  

1. The **average capital gain** across all individuals is approximately **1,079**.
   But this is misleading without distribution context (explained below).

2. The **standard deviation** is **7,452**, which indicates **high variability** in capital gains — some people have very high gains while most have none.

3. The **minimum capital gain** is **0** — many people didn’t report any capital gain.

4. The **maximum capital gain** is **99,999**, showing at least one person had a very large gain. This is an extreme outlier.

5. The **25th percentile** is **0**, meaning **25% of the people had zero capital gain**.

6. The **50th percentile (median)** is **0**, so **at least 50% of individuals had no capital gain at all**.

7. The **75th percentile** is also **0**, meaning **at least 75% of the people had zero capital gain**.

Capital Gain is **heavily right-skewed**. Most people had no capital gain, and only a small number had very high values. So:

* The **mean is inflated** by outliers.
* Use **histograms** or **log scale plots** to visualize this skewed distribution.
* Consider analyzing **non-zero gains separately**.

**Capital Loss:**  

1. The **average capital loss** is about **87.5**, but again, that’s not very meaningful due to skewness.

2. The **standard deviation** is **403**, indicating large variability.

3. The **minimum value** is **0**, meaning many individuals had no capital loss.

4. The **maximum capital loss** is **4,356**.

5. The **25th percentile** is **0**, so **25% of the people had zero loss**.

6. The **median** is also **0**, so **at least 50% of individuals had no capital loss**.

7. The **75th percentile** is **0** again, meaning **at least 75% had no capital loss**.

Similar to capital gain, **capital loss is also heavily right-skewed**. Most people didn’t report any losses, and a few had high losses. This again inflates the mean.
