### **1. Relationship Status Frequency:**

```python
aa= data['relationship'].value_counts().reset_index()
aa
```

```python
plt.figure(figsize=(15, 4))
sns.set_theme(context= "notebook", style= "dark")
sns.barplot(data= aa, x= "relationship", y= "count", palette= "magma", hue= "relationship", legend= False)
plt.title("Relationship Status Frequency:", fontsize= 15, fontweight= "bold", color= "Red")
plt.xlabel("(Relationship)", color= "Purple", fontweight= "bold")
plt.ylabel("(Count)", color= "Purple", fontweight= "bold")
plt.show()
```

The plot displays the frequency of different relationship statuses, highlighting that "Husband" has the highest count, exceeding 16,000 individuals. "Not-in-family" follows as the second most common status, while "Own-child," "Unmarried," "Wife," and "Other-relative" show significantly lower frequencies. This distribution suggests that traditional family structures, particularly the role of husbands, are prevalent in the dataset, while other relationship types are less represented.

### **4. Single vs Married: Income Distribution:**

```python
data.columns
```

```python
data.head(3)
```

```python
data[["marital-status", "salary"]].nunique()
```

```python
data["marital-status"].unique()
```

```python
mms= data.groupby(by= "marital-status")["salary"].value_counts().sort_values(ascending= False).reset_index()
mms
```

```python
# Lists all available seaborn palettes: 
print(sns.palettes.SEABORN_PALETTES.keys())
```

```python
plt.figure(figsize=(15, 4))
sns.set_theme(context= "notebook", style= "dark")
sns.barplot(data= mms, hue= "marital-status", y= "count", x= "salary", palette= "bright6")
plt.title("Marital Status and Income:", fontsize= 15, fontweight= "bold", color= "Red")
plt.xlabel("Salary", fontsize= 12, fontweight= "bold", color= "purple")
plt.ylabel("Counts", fontsize= 12, fontweight= "bold", color= "purple")
plt.show()
```

The main takeaway is that people who are married to civilian spouses ("Married-civ-spouse") are much more likely to earn higher salaries (>50K), while those who are never married, divorced, separated, widowed, or in non-traditional marital arrangements (like "Married-spouse-absent" or "Married-AF-spouse") are more commonly found in the lower income bracket (≤50K). 

In `<=50k` income bracket, people who are `not married` (not even once) are the most in number followed by `married to civilian spouses` and `divorsed`. So, low incoome bracket is mainly dominated by people from `unstable-relation` status whereas, high income group is highly dominated by `stable-relation-group` particularly `married to civilian spouses`.

### **5. Working Hours vs Marital Status:**

```python
data.columns
```

```python
data.head(2)
```

```python
data[["marital-status", "hours-per-week"]]
```

```python
data["marital-status"].unique()
```

```python
data["hours-per-week"].unique()
```

```python
sas= data.groupby(by= "marital-status")["hours-per-week"].mean().reset_index()
sas
```

```python
plt.figure(figsize=(15, 4))
sns.set_theme(context= "notebook", style= "dark")
sns.barplot(data= sas, x= "marital-status", y= "hours-per-week", palette= "bright6")
plt.title("Marital Status vs Hours-per-week:", fontsize= 15, fontweight= "bold", color= "Red")
plt.xlabel("Marital Status", fontsize= 12, fontweight= "bold", color= "purple")
plt.ylabel("Average Work Hours-per-week", fontsize= 12, fontweight= "bold", color= "purple")
# add annotations: 
for i in range(len(sas)):
    plt.annotate(round(sas["hours-per-week"][i], 2), xy=(i, sas["hours-per-week"][i]), ha='center', va='bottom', color= "Gray", fontweight= "bold")
plt.show()
```

The bar chart shows the average number of work hours per week across different marital statuses. **Married-civ-spouse** individuals work the most on average (43.46 hours/week), followed by **Divorced** (41.16) and **Separated** (39.67). 

**Widowed** individuals work the least, averaging just **33.45 hours/week**. The plot suggests that marital status may be associated with work engagement, with married or previously married individuals generally working more than those who are never married or widowed. 

```python
data["marital-status"].unique() 
```

```python
plt.figure(figsize=(15, 4))
sns.set_theme(context= "notebook", style= "dark")
sns.boxplot(data= data, x= "marital-status", y= "hours-per-week", palette= "bright6", width= 0.3, )
plt.title("Marital Status vs Hours-per-week:", fontsize= 15, fontweight= "bold", color= "Red")
plt.xlabel("Marital Status", fontsize= 12, fontweight= "bold", color= "purple")
plt.ylabel("Work Hours-per-week", fontsize= 12, fontweight= "bold", color= "purple")
plt.xticks(color= "Gray", fontweight= "bold")
plt.yticks(color= "Gray", fontweight= "bold")
plt.show()
```

The boxplot displays the distribution of weekly work hours across different marital statuses. Most groups have a median around 40 hours, but the **Widowed** category shows a lower median and wider spread, suggesting more variability and fewer consistent full-time workers. 

**Divorced** and **Married-civ-spouse** individuals tend to have slightly higher medians and tighter distributions, indicating more stable full-time work patterns. 

All groups show a large number of outliers above the upper whisker, with some individuals reporting over 60–80 hours per week. This suggests that while average work hours cluster near 40, a notable portion of individuals work significantly more regardless of marital status.
